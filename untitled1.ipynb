```python
from IPython.display import display, HTML

display(HTML("""
<!DOCTYPE html>
<html lang="ko">
<head>
    <meta charset="UTF-8">
    <title>Flexbox - After</title>

    <style>
        body {
            font-family: Arial, sans-serif;
            margin: 20px;
            color: #18233f;
        }

        .title {
            font-size: 18px;
            font-weight: bold;
            margin-bottom: 20px;
        }

        .content {
            display: flex;
            gap: 18px;
        }

        .code-area {
            width: 50%;
        }

        .result-area {
            width: 50%;
            min-height: 280px;
            border: 1px solid #d6dce5;
            border-radius: 10px;
            padding: 20px;
        }

        .code-box {
            background-color: #111d39;
            color: white;
            padding: 20px;
            border-radius: 10px;
            font-family: Consolas, monospace;
            line-height: 1.8;
            min-height: 240px;
        }

        .browser-title {
            color: #8a94a6;
            font-size: 13px;
            margin-bottom: 15px;
        }

        /* 부모에게 flex 적용 */
        .container {
            display: flex;
            justify-content: center;
            gap: 12px;
        }

        .box {
            width: 55px;
            padding: 10px 5px;
            border-radius: 6px;
            color: white;
            text-align: center;
            font-size: 13px;
        }

        .box1 {
            background-color: #16a6c1;
        }

        .box2 {
            background-color: #08758b;
        }

        .box3 {
            background-color: #111d39;
        }

        .key-point {
            margin-top: 15px;
            padding: 15px 20px;
            background-color: #eaf7fa;
            border-left: 4px solid #16a6c1;
            line-height: 1.7;
        }
    </style>
</head>

<body>

    <div class="title">
        After — 부모(container)에 display: flex 적용
    </div>

    <div class="content">

        <!-- 왼쪽 코드 -->
        <div class="code-area">
            <div class="code-box">
&lt;!-- HTML: 박스 3개를 감싸는 부모 --&gt;

&lt;div class="container"&gt;
    &lt;div&gt;박스1&lt;/div&gt;
    &lt;div&gt;박스2&lt;/div&gt;
    &lt;div&gt;박스3&lt;/div&gt;
&lt;/div&gt;


&lt;!-- CSS: 부모에게 flex 지시 --&gt;

.container {
    display: flex;
    justify-content: center;
    gap: 12px;
}
            </div>
        </div>

        <!-- 오른쪽 결과 -->
        <div class="result-area">

            <div class="browser-title">
                브라우저 화면
            </div>

            <div class="container">
                <div class="box box1">박스1</div>
                <div class="box box2">박스2</div>
                <div class="box box3">박스3</div>
            </div>

        </div>

    </div>

    <div class="key-point">
        <strong>핵심:</strong>
        <code>display: flex</code>는 자식(박스1, 2, 3)이 아니라
        부모(<code>container</code>)에게 주는 명령입니다.
        부모에게 flex를 적용하면 자식 요소들이 기본적으로 가로로 나란히 배치됩니다.
    </div>

</body>
</html>
"""))
```
